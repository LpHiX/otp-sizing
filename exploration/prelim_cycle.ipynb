```python
import numpy as np

F = 25000
p_c_bar = 80
p_e_bar = 0.8
p_a_bar = 1
# Temporary variables
gamma = 1.2
esp = 10


def compute_CF(Pc, Pe, Pa, gamma, esp):
    """
    Compute the coefficient of thrust (CF) for a rocket engine.

    Parameters:
    Pc : float
        Chamber pressure (Pa / bar)
    Pe : float
        Exit pressure (Pa / bar)
    Pa : float
        Ambient pressure (Pa / bar)
    gamma : float
        Specific heat ratio
    esp : float
        Expansion ratio

    Returns:
    float
        Coefficient of thrust (CF)
    """
    return np.sqrt((2 * gamma * gamma) / ((gamma - 1) * (2 / (gamma + 1)) ** ((gamma + 1) / (gamma - 1)) * (1 - (Pe / Pc) ** ((gamma - 1) / gamma)))) + (Pe - Pa) / Pc * esp 

print(compute_CF(p_c_bar, p_e_bar, p_a_bar, gamma, esp))
```
